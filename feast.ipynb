```python
import pandas as pd
df = pd.read_csv("/content/skill_gap.csv")
technical_cols = [
    "python",
    "java",
    "sql",
    "dsa",
    "machine_learning",
    "cloud",
    "web_development"
]
df["technical_skill_score"] = df[technical_cols].mean(axis=1)
soft_cols = ["communication", "problem_solving"]

df["soft_skill_score"] = df[soft_cols].mean(axis=1)
df["strong_skill_count"] = (
    df[technical_cols + soft_cols] >= 7
).sum(axis=1)
df["weak_skill_count"] = (
    df[technical_cols + soft_cols] < 5
).sum(axis=1)
df["overall_skill_score"] = (
    df["technical_skill_score"] * 0.7 +
    df["soft_skill_score"] * 0.3
)
print(df.head())
df.to_csv("skill_gap_features.csv", index=False)
print("Feature extraction completed successfully!")
```

```python
base_time = pd.Timestamp(
    "2024-01-01",
    tz="UTC"
)
df["event_timestamp"] = (
    base_time +
    pd.to_timedelta(
        df["student_id"].str.extract(r'(\d+)')[0].astype(int),
        unit="s"
    )
)
df["created_timestamp"] = (
    df["event_timestamp"] +
    pd.Timedelta(seconds=1)
)
df[
    [
        "student_id",
        "event_timestamp",
        "created_timestamp"
    ]
].head()
```

```python
import os

repo_path = "/content/skill_gap_feast"

os.makedirs(
    f"{repo_path}/data",
    exist_ok=True
)
```

```python
feature_df.to_parquet(
    f"{repo_path}/data/skill_gap_features.parquet",
    index=False
)
```

```python
feature_store_yaml = """
project: skill_gap_project

registry: data/registry.db

provider: local

offline_store:
  type: file

online_store:
  type: sqlite
  path: data/online_store.db
"""

with open(
    f"{repo_path}/feature_store.yaml",
    "w"
) as f:
    f.write(feature_store_yaml)
```

```python
feature_definition = '''
from datetime import timedelta

from feast import (
    Entity,
    FeatureView,
    FeatureService,
    Field,
    FileSource
)

from feast.types import (
    Float32,
    Int64
)


# -----------------------------
# ENTITY
# -----------------------------

student = Entity(
    name="student",
    join_keys=["student_id"],
    description="Student in the curriculum-industry skill-gap dataset"
)


# -----------------------------
# DATA SOURCE
# -----------------------------

skill_gap_source = FileSource(
    name="skill_gap_source",
    path="data/skill_gap_features.parquet",
    timestamp_field="event_timestamp",
    created_timestamp_column="created_timestamp"
)


# -----------------------------
# FEATURE VIEW
# -----------------------------

skill_gap_feature_view = FeatureView(
    name="skill_gap_features",
    entities=[student],

    ttl=timedelta(days=50000),

    schema=[
        Field(
            name="technical_skill_score",
            dtype=Float32
        ),

        Field(
            name="soft_skill_score",
            dtype=Float32
        ),

        Field(
            name="strong_skill_count",
            dtype=Int64
        ),

        Field(
            name="weak_skill_count",
            dtype=Int64
        ),

        Field(
            name="overall_skill_score",
            dtype=Float32
        ),
    ],

    source=skill_gap_source,

    online=True
)


# -----------------------------
# FEATURE SERVICE
# -----------------------------

skill_gap_feature_service = FeatureService(
    name="skill_gap_service",
    features=[
        skill_gap_feature_view
    ]
)
'''

with open(
    f"{repo_path}/features.py",
    "w"
) as f:
    f.write(feature_definition)

print("features.py created successfully!")
```

```python
from feast import FeatureStore

store = FeatureStore(
    repo_path="/content/skill_gap_feast"
)
```

```python
feature_service = store.get_feature_service(
    "skill_gap_service"
)
```

```python
entity_df = label_df.copy()

display(entity_df.head())
```

```python
training_data = store.get_historical_features(
    entity_df=entity_df,
    features=feature_service
).to_df()
```

```python
display(training_data.head())
```

```python
feature_columns = [
    "python",
    "java",
    "sql",
    "dsa",
    "machine_learning",
    "cloud",
    "web_development",
    "communication",
    "problem_solving",
    "industry_skill_score",
    "skill_gap_score",
    "industry_readiness_score",
    "average_skill_score",
    "strong_skill_count",
    "weak_skill_count",
    "technical_skill_score"
]
```

```python
feature_columns = [
    "technical_skill_score",
    "soft_skill_score",
    "strong_skill_count",
    "weak_skill_count",
    "overall_skill_score"
]

X = training_data[feature_columns]

y = training_data["placement"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

model.fit(
    X_train,
    y_train
)
```

```python
predictions = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(
    y_test,
    predictions
)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```
